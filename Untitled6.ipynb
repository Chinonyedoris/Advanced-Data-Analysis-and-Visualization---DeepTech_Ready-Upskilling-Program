```python
age_slider = pn.widgets.IntRangeSlider(name='Age', start=df['Age'].min(), end=df['Age'].max(), step=1, value=(df['Age'].min(), df['Age'].max()))
gender_select = pn.widgets.Select(name='Gender', options=list(df['Gender'].unique()))
smoking_select = pn.widgets.Select(name='Smoking Status', options=list(df['Smoking_Status'].unique()))
alcohol_select = pn.widgets.Select(name='Alcohol Intake', options=list(df['Alcohol_Intake'].unique()))
physical_activity_select = pn.widgets.Select(name='Physical Activity', options=list(df['Physical_Activity'].unique()))
diet_quality_select = pn.widgets.Select(name='Diet Quality', options=list(df['Diet_Quality'].unique()))

print("Interactive widgets defined.")
```

## Implement Data Filtering

### Subtask:
Create a function to dynamically filter the dataset based on the selections made through the interactive widgets. This filtered data will be used for both visualization and export.


```python
def filter_data(age_range, gender, smoking_status, alcohol_intake, physical_activity, diet_quality):
    filtered_df = df[
        (df['Age'].between(age_range[0], age_range[1])) &
        (df['Gender'] == gender) &
        (df['Smoking_Status'] == smoking_status) &
        (df['Alcohol_Intake'] == alcohol_intake) &
        (df['Physical_Activity'] == physical_activity) &
        (df['Diet_Quality'] == diet_quality)
    ]
    return filtered_df

@pn.depends(
    age_slider.param.value,
    gender_select.param.value,
    smoking_select.param.value,
    alcohol_select.param.value,
    physical_activity_select.param.value,
    diet_quality_select.param.value
)
def reactive_filtered_df(age_range, gender, smoking_status, alcohol_intake, physical_activity, diet_quality):
    return filter_data(age_range, gender, smoking_status, alcohol_intake, physical_activity, diet_quality)

print("Data filtering function and reactive DataFrame definition complete.")
```

## Generate Bar Chart

### Subtask:
Create a bar chart using the filtered data. The chart will display the distribution of 'Heart_Attack_Outcome' (0 = No, 1 = Yes). The x-axis will represent 'Heart Attack Outcome', and the y-axis will show the count of patients in each category. The chart title will be 'Filtered Heart Attack Outcome Distribution (Bar Chart)'.


```python
def plot_heart_attack_outcome(df_filtered):
    if df_filtered.empty:
        return pn.pane.Markdown("### No data matches the current filters.")

    # Calculate the distribution of 'Heart_Attack_Outcome'
    outcome_counts = df_filtered['Heart_Attack_Outcome'].value_counts().reset_index()
    outcome_counts.columns = ['Heart_Attack_Outcome', 'Count']

    # Map 0 to 'No' and 1 to 'Yes' for better readability
    outcome_counts['Heart_Attack_Outcome_Label'] = outcome_counts['Heart_Attack_Outcome'].map({0: 'No', 1: 'Yes'})

    # Create the bar chart using Altair
    chart = alt.Chart(outcome_counts).mark_bar().encode(
        x=alt.X('Heart_Attack_Outcome_Label:N', title='Heart Attack Outcome'),
        y=alt.Y('Count:Q', title='Number of Patients'),
        tooltip=['Heart_Attack_Outcome_Label', 'Count']
    ).properties(
        title='Filtered Heart Attack Outcome Distribution (Bar Chart)'
    ).interactive()

    return chart

# Make the plot function reactive to the filtered DataFrame
reactive_heart_attack_chart = pn.bind(plot_heart_attack_outcome, df_filtered=reactive_filtered_df)

print("Bar chart generation function and reactive chart definition complete.")
```

## Create Data Export Functionality

### Subtask:
Add a button to the dashboard that, when clicked, allows the user to download the currently filtered dataset as a CSV file.


```python
def export_csv_button(df_filtered):
    if df_filtered.empty:
        return pn.pane.Markdown("### No data to export.")

    # Convert the filtered DataFrame to a CSV string
    csv_data = df_filtered.to_csv(index=False)

    # Create a FileDownload widget
    download_button = pn.widgets.FileDownload(
        file=csv_data.encode('utf-8'), # FileDownload expects bytes
        filename='filtered_heart_attack_data.csv',
        button_type='success',
        label='Download Filtered Data (CSV)'
    )
    return download_button

# Make the download button reactive to the filtered DataFrame
reactive_download_button = pn.bind(export_csv_button, df_filtered=reactive_filtered_df)

print("Data export button functionality defined and made reactive.")
```

```python
dashboard = pn.Column(
    "# Heart Attack Outcome Dashboard",
    pn.Row(
        pn.Column(
            "## Filters",
            age_slider,
            gender_select,
            smoking_select,
            alcohol_select,
            physical_activity_select,
            diet_quality_select,
            reactive_download_button
        ),
        pn.Column(
            "## Heart Attack Outcome Distribution",
            reactive_heart_attack_chart
        )
    )
)

dashboard.servable()

print("Dashboard layout created and made servable.")
```
