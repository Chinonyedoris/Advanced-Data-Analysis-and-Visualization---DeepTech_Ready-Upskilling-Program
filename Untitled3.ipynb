## Load Dataset

### Subtask:
Load the `job_change_data.csv` file into a pandas DataFrame for analysis.


````python
import pandas as pd

df = pd.read_csv('/content/job_change_data.csv')
df.head()
````

## Filter Rows by Conditions

### Subtask:
Filter the DataFrame to include only rows where 'City_Development_Index' > 0.8 and 'Company_Size' > 3.


````python
company_size_mapping = {
    '<10': 0,
    '10-49': 10,
    'Oct-49': 10,  # Handling the typo as specified
    '50-99': 50,
    '100-500': 100,
    '500-999': 500,
    '1000-4999': 1000,
    '5000-9999': 5000,
    '10000+': 10000
}

df['company_size_numeric'] = df['company_size'].map(company_size_mapping)

df_filtered = df[
    (df['city_development_index'] > 0.8) &
    (df['company_size_numeric'] > 3)
]

print("First 5 rows of the filtered DataFrame:")
print(df_filtered.head())
print("\nShape of the filtered DataFrame:")
print(df_filtered.shape)
````

## Select Specific Data with iloc

### Subtask:
Use `iloc` to select the first 10 rows and the 'Experience' and 'Education_Level' columns from the filtered DataFrame.


````python
df_selected = df_filtered.iloc[0:10, df_filtered.columns.get_indexer(['experience', 'education_level'])]
print("Selected DataFrame (first 10 rows, 'experience' and 'education_level' columns):")
print(df_selected)
````

## Group by Relevant_Experience

### Subtask:
Group the data by 'Relevant_Experience' and calculate the average 'City_Development_Index' for each group.


````python
df['experience_numeric'] = df['experience'].replace({'<1': '0', '>20': '21'}).astype(str)
df['experience_numeric'] = pd.to_numeric(df['experience_numeric'], errors='coerce')

experience_city_development_index_mean = df.groupby('relevent_experience')['city_development_index'].mean()

print("Average 'City_Development_Index' by 'Relevant_Experience':")
print(experience_city_development_index_mean)
````

## Group by Relevant_Experience and Company_Size

### Subtask:
Group the data by 'Relevant_Experience' and 'Company_Size' to calculate the average 'City_Development_Index' and the number of unique 'enrollee_id' for each group.

#### Instructions
1. Group the DataFrame `df` by both 'relevent_experience' and 'company_size'.
2. For each group, calculate the mean of 'city_development_index' and the number of unique 'enrollee_id'.
3. Display the resulting grouped data.

````python
grouped_data = df.groupby(['relevent_experience', 'company_size']).agg(
    avg_city_development_index=('city_development_index', 'mean'),
    unique_enrollee_ids=('enrollee_id', 'nunique')
)

print("Grouped data by 'Relevant_Experience' and 'Company_Size':")
print(grouped_data.head())
````

## Analyze Company_Type Frequency

## Handle Missing Numerical Values

### Subtask:
Handle missing numerical values in the DataFrame by filling them with the mean of their respective columns.

#### Instructions
1. Identify numerical columns in the DataFrame that contain missing values.
2. For each identified numerical column, calculate its mean.
3. Fill the missing values in these numerical columns with their respective means.
4. Display the count of missing values for the numerical columns after filling.

````python
numerical_cols = df.select_dtypes(include=['number']).columns

for col in numerical_cols:
    if df[col].isnull().any():
        mean_val = df[col].mean()
        df[col].fillna(mean_val, inplace=True)
        print(f"Filled missing values in '{col}' with mean: {mean_val:.2f}")

print("\nMissing values after filling (numerical columns only):")
print(df[numerical_cols].isnull().sum())
````

## Drop Rows with Excessive Missing Data

### Subtask:
Drop rows from the DataFrame that have more than 50% of their values missing.

#### Instructions
1. Calculate the threshold for missing values: 50% of the total number of columns.
2. Drop rows where the number of non-null values is less than this threshold.
3. Display the shape of the DataFrame before and after dropping rows, and show the first 5 rows of the modified DataFrame.

````python
print(f"Shape of DataFrame before dropping rows: {df.shape}")

# Calculate the threshold for non-null values (50% of total columns)
threshold = len(df.columns) * 0.5

# Drop rows where the number of non-null values is less than the threshold
df_dropped_missing = df.dropna(thresh=threshold)

print(f"Shape of DataFrame after dropping rows: {df_dropped_missing.shape}")
print("\nFirst 5 rows of the DataFrame after dropping rows with excessive missing data:")
print(df_dropped_missing.head())
````

## Create an Experience_Gap Feature

### Subtask:
Create a new feature called 'Experience_Gap' by calculating the difference between 'experience_numeric' and a hypothetical 'ideal_experience' of 10 years.

#### Instructions
1. Define a constant for 'ideal_experience' (e.g., 10).
2. Calculate 'Experience_Gap' as `experience_numeric - ideal_experience`.
3. Display the first 5 rows of the DataFrame including the new feature.


````python
ideal_experience = 10
df_dropped_missing['Experience_Gap'] = df_dropped_missing['experience_numeric'] - ideal_experience

print("First 5 rows of DataFrame with new 'Experience_Gap' feature:")
print(df_dropped_missing.head())
````

## Normalize City_Development_Index

### Subtask:
Normalize the 'City_Development_Index' column using Min-Max scaling to a range between 0 and 1.

#### Instructions
1. Calculate the minimum and maximum values of the 'city_development_index' column.
2. Apply Min-Max normalization using the formula: `(value - min) / (max - min)`.
3. Create a new column, 'City_Development_Index_Normalized', to store the normalized values.
4. Display the first 5 rows of the DataFrame, including the new normalized column, and also print the minimum and maximum values of the normalized column to verify the scaling.

## Normalize City_Development_Index

### Subtask:
Normalize the 'City_Development_Index' column using Min-Max scaling to a range between 0 and 1.


````python
min_cdi = df_dropped_missing['city_development_index'].min()
max_cdi = df_dropped_missing['city_development_index'].max()

df_dropped_missing['City_Development_Index_Normalized'] = (df_dropped_missing['city_development_index'] - min_cdi) / (max_cdi - min_cdi)

print("First 5 rows of DataFrame with 'City_Development_Index_Normalized' feature:")
print(df_dropped_missing.head())

print("\nMin of 'City_Development_Index_Normalized':", df_dropped_missing['City_Development_Index_Normalized'].min())
print("Max of 'City_Development_Index_Normalized':", df_dropped_missing['City_Development_Index_Normalized'].max())
````

## Create and Merge 'cdi_per' Column

### Subtask:
Create a new DataFrame `city_development_performance` by grouping the original DataFrame by 'city' and calculating the mean of the 'City_Development_Index_Normalized' column. Then, merge this new `cdi_per` column into the main DataFrame.

#### Instructions
1. Group the `df_dropped_missing` DataFrame by 'city' and calculate the mean of 'City_Development_Index_Normalized'.
2. Rename the resulting mean column to 'cdi_per'.
3. Merge this new `city_development_performance` DataFrame back into `df_dropped_missing` using 'city' as the key.
4. Display the first 5 rows of the `df_dropped_missing` DataFrame with the newly merged 'cdi_per' column.

````python
city_development_performance = df_dropped_missing.groupby('city')['City_Development_Index_Normalized'].mean().reset_index()
city_development_performance.rename(columns={'City_Development_Index_Normalized': 'cdi_per'}, inplace=True)

df_dropped_missing = pd.merge(df_dropped_missing, city_development_performance, on='city', how='left')

print("First 5 rows of DataFrame with newly merged 'cdi_per' column:")
print(df_dropped_missing.head())
````
