## Load All Worksheets

### Subtask:
Load all worksheets from the 'Super Store Normalized.xlsx' Excel workbook into separate pandas DataFrames to represent the individual tables.


```python
import pandas as pd
print("pandas imported successfully.")
```

```python
excel_file = pd.ExcelFile('/content/Super Store Normalized.xlsx')
sheet_names = excel_file.sheet_names

dataframes = {}
for sheet in sheet_names:
    df = excel_file.parse(sheet)
    dataframes[sheet] = df
    print(f"Loaded sheet: {sheet} into a DataFrame")

print("\nAll loaded DataFrame names:")
for sheet_name in dataframes.keys():
    print(sheet_name)
```

## Examine Data Structure and Relationships

### Subtask:
Analyze the loaded DataFrames to understand their structure, identify potential primary and foreign keys, and describe the relationships between the tables. This will help in understanding the existing data model.


```python
for name, df in dataframes.items():
    print(f"\n--- DataFrame: {name} ---")
    print("Structure Info:")
    df.info()
    print("\nFirst 5 rows:")
    print(df.head())
    print("\nDescriptive Statistics:")
    print(df.describe(include='all'))
```

## Identify and Combine Related Dimension Tables

### Subtask:
Determine which tables are dimension tables and if any of them are related to each other. If related, combine these dimension tables into a single, comprehensive dimension table where appropriate, as if preparing to disable unused tables from loading in Power BI.


```python
dim_locations = pd.merge(dataframes['Locations'], dataframes['States'], on='State ID', how='left')
print("Created dim_locations by merging Locations and States.")

dim_products = pd.merge(dataframes['Products'], dataframes['Products Pricing Tiers'], on='Pricing Code', how='left')
print("Created dim_products by merging Products and Products Pricing Tiers.")

dim_customers = dataframes['Customers'].copy()
print("Renamed Customers to dim_customers.")

print("\n--- Verifying dim_locations ---")
print(dim_locations.head())
dim_locations.info()

print("\n--- Verifying dim_products ---")
print(dim_products.head())
dim_products.info()

print("\n--- Verifying dim_customers ---")
print(dim_customers.head())
dim_customers.info()
```

## Remove Unnecessary Columns

### Subtask:
Identify and remove columns that are not required for the model or reports, specifically excluding 'Order ID' from all relevant DataFrames, and any other columns identified as redundant during the data examination phase.


```python
dataframes['Orders'] = dataframes['Orders'].drop('Order ID', axis=1)
print("Removed 'Order ID' from Orders DataFrame.")

dim_locations = dim_locations.drop('State ID', axis=1)
print("Removed 'State ID' from dim_locations DataFrame.")

dim_products = dim_products.drop('Pricing Code', axis=1)
print("Removed 'Pricing Code' from dim_products DataFrame.")

print("\n--- Verifying Orders DataFrame ---")
print(dataframes['Orders'].info())
print(dataframes['Orders'].head())

print("\n--- Verifying dim_locations DataFrame ---")
print(dim_locations.info())
print(dim_locations.head())

print("\n--- Verifying dim_products DataFrame ---")
print(dim_products.info())
print(dim_products.head())
```

## Summarize Optimized Data Model

### Subtask:
Display the structure (e.g., columns, first few rows, and data types) of the resulting DataFrames after combining tables and removing columns, highlighting the optimized data model for Power BI.


```python
print("\n--- Optimized Data Model Summary ---")

print("\n--- Fact Table: Orders ---")
print(dataframes['Orders'].info())
print(dataframes['Orders'].head())

print("\n--- Dimension Table: Customers ---")
print(dim_customers.info())
print(dim_customers.head())

print("\n--- Dimension Table: Locations ---")
print(dim_locations.info())
print(dim_locations.head())

print("\n--- Dimension Table: Products ---")
print(dim_products.info())
print(dim_products.head())
```

## Visualize Relationships and Key Metrics

### Subtask:
Create a visualization (e.g., a bar chart or line plot) that leverages data from different 'tables' (DataFrames) to demonstrate how relationships would function and to show key metrics, mimicking the testing of model relationships with dimension and fact fields in Power BI. Ensure to include legends for clarity.


```python
import matplotlib.pyplot as plt
import seaborn as sns
print("matplotlib.pyplot and seaborn imported successfully.")
```

```python
orders_products = pd.merge(dataframes['Orders'], dim_products, on='Product ID', how='left')
print("Merged Orders and Products data.")

final_df = pd.merge(orders_products, dim_locations, on='Location ID', how='left')
print("Merged Orders-Products with Locations data.")

sales_by_category_region = final_df.groupby(['Category', 'Region'])['Sales'].sum().reset_index()
print("Grouped data by Category and Region and summed Sales.")

plt.figure(figsize=(12, 7))
sns.barplot(x='Category', y='Sales', hue='Region', data=sales_by_category_region)
plt.title('Total Sales by Product Category and Region')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.legend(title='Region')
plt.tight_layout()
plt.show()
print("Bar plot showing Total Sales by Product Category and Region displayed.")
```
